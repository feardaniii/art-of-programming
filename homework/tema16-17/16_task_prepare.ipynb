### Exercițiul 2 – Selectează rapid ofertele premium
Folosește DataFrame-ul din exercițiul anterior pentru a extrage doar produsele cu `Preț` mai mare de 50. Compare filtrarea cu mască booleană cu o abordare `loc` sau `.query()`, astfel încât să vezi diferențele de sintaxă și expresivitate.


```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Create DataFrame from Exercise 1
produse = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam']
preturi = np.random.randint(50, 3000, size=5)
stocuri = np.random.randint(5, 50, size=5)

df = pd.DataFrame({
    'Produs': produse,
    'Pret': preturi,
    'Stoc': stocuri
})

print("="*60)
print("DATAFRAME ORIGINAL")
print("="*60)
print(df)

# ==================== METHOD 1: BOOLEAN MASKING ====================
print("\n" + "="*60)
print("METODA 1: Boolean Masking")
print("="*60)

# Create mask
mask = df['Pret'] > 50
print("\nMasca (True/False pentru fiecare rând):")
print(mask)

# Apply mask
produse_premium_v1 = df[mask]
print("\nProduse premium (Preț > 50):")
print(produse_premium_v1)

# Or in one line:
produse_premium_v1_short = df[df['Pret'] > 50]

# ==================== METHOD 2: .loc[] ====================
print("\n" + "="*60)
print("METODA 2: Folosind .loc[]")
print("="*60)

produse_premium_v2 = df.loc[df['Pret'] > 50]
print(produse_premium_v2)

# With column selection
print("\nCu selectare de coloane (doar Produs și Preț):")
produse_premium_v2_cols = df.loc[df['Pret'] > 50, ['Produs', 'Pret']]
print(produse_premium_v2_cols)

# ==================== METHOD 3: .query() ====================
print("\n" + "="*60)
print("METODA 3: Folosind .query()")
print("="*60)

produse_premium_v3 = df.query('Pret > 50')
print(produse_premium_v3)

# Using variable
threshold = 50
produse_premium_v3_var = df.query('Pret > @threshold')
print(f"\nFolosind variabilă (threshold = {threshold}):")
print(produse_premium_v3_var)

# ==================== COMPARISON ====================
print("\n" + "="*60)
print("COMPARAȚIE SINTAXĂ")
print("="*60)

print("\n1. Boolean Masking:")
print("   df[df['Pret'] > 50]")
print("   ✓ Cel mai comun")
print("   ✓ Direct și simplu")
print("   ✗ Poate deveni greu de citit cu condiții complexe")

print("\n2. .loc[]:")
print("   df.loc[df['Pret'] > 50]")
print("   ✓ Mai explicit")
print("   ✓ Permite selectare ușoară de coloane")
print("   ✓ Recomandat când vrei și să selectezi coloane")

print("\n3. .query():")
print("   df.query('Pret > 50')")
print("   ✓ Sintaxă foarte curată (SQL-like)")
print("   ✓ Excelent pentru condiții multiple complexe")
print("   ✓ Mai ușor de citit pentru non-programatori")

# ==================== VERIFICATION ====================
print("\n" + "="*60)
print("VERIFICARE: Toate metodele dau același rezultat?")
print("="*60)

print(f"Metoda 1: {len(produse_premium_v1)} produse")
print(f"Metoda 2: {len(produse_premium_v2)} produse")
print(f"Metoda 3: {len(produse_premium_v3)} produse")
print(f"\n✓ Toate metodele returnează același număr de produse!")
```

### Exercițiul 3 – Pregătește ordinea pentru analize ulterioare
Continuăm pregătirea datasetului pentru vizualizările din modulul 18-19. Sortează produsele descrescător după `Preț` cu `sort_values()` și afișează primele trei intrări pentru a verifica rezultatul.


```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Create DataFrame from previous exercises
produse = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam']
preturi = np.random.randint(50, 3000, size=5)
stocuri = np.random.randint(5, 50, size=5)

df = pd.DataFrame({
    'Produs': produse,
    'Pret': preturi,
    'Stoc': stocuri
})

print("="*60)
print("DATAFRAME ORIGINAL (nesortat)")
print("="*60)
print(df)
print(f"\nPrețuri în ordine actuală: {df['Pret'].tolist()}")

# ==================== SORTING ====================
print("\n" + "="*60)
print("SORTARE DUPĂ PREȚ (descrescător)")
print("="*60)

# Sort by price (descending = highest first)
df_sortat = df.sort_values(by='Pret', ascending=False)

print("\nDataFrame sortat (toate produsele):")
print(df_sortat)
print(f"\nPrețuri după sortare: {df_sortat['Pret'].tolist()}")

# ==================== TOP 3 ====================
print("\n" + "="*60)
print("TOP 3 CELE MAI SCUMPE PRODUSE")
print("="*60)

top_3 = df_sortat.head(3)
print(top_3)

# Alternative: sort and get top 3 in one line
print("\n" + "="*60)
print("ALTERNATIVĂ: Sort și head() într-o singură linie")
print("="*60)
top_3_oneliner = df.sort_values(by='Pret', ascending=False).head(3)
print(top_3_oneliner)

# ==================== VERIFICATION ====================
print("\n" + "="*60)
print("VERIFICARE")
print("="*60)

print(f"✓ Număr total produse: {len(df)}")
print(f"✓ Top 3 afișate: {len(top_3)}")
print(f"✓ Cel mai scump produs: {top_3.iloc[0]['Produs']} - {top_3.iloc[0]['Pret']} RON")
print(f"✓ Al doilea ca preț: {top_3.iloc[1]['Produs']} - {top_3.iloc[1]['Pret']} RON")
print(f"✓ Al treilea ca preț: {top_3.iloc[2]['Produs']} - {top_3.iloc[2]['Pret']} RON")

# Verify descending order
print(f"\n✓ Verificare ordine descrescătoare:")
preturi_top3 = top_3['Pret'].tolist()
if preturi_top3 == sorted(preturi_top3, reverse=True):
    print(f"   Corect! {preturi_top3[0]} >= {preturi_top3[1]} >= {preturi_top3[2]}")
else:
    print("   ✗ Eroare la sortare!")

# ==================== BONUS: Other sorting examples ====================
print("\n" + "="*60)
print("BONUS: Alte exemple de sortare")
print("="*60)

# Sort ascending (cheapest first)
print("\nCele mai ieftine 3 produse (crescător):")
top_3_ieftine = df.sort_values(by='Pret', ascending=True).head(3)
print(top_3_ieftine)

# Sort by stock
print("\nProduse sortate după stoc (descrescător):")
sortat_stoc = df.sort_values(by='Stoc', ascending=False)
print(sortat_stoc)

# Sort by multiple columns
print("\nSortare după Preț apoi Stoc:")
sortat_multiplu = df.sort_values(by=['Pret', 'Stoc'], ascending=[False, False])
print(sortat_multiplu)
```

### Exercițiul 4 – Calculează automat TVA-ul
Adaugă o coloană `TVA` ce reprezintă 19% din prețul fiecărui produs și păstrează două zecimale prin rotunjire vectorizată. Observă cum Pandas aplică operațiile pe tot coloana și confirmă cu `head()` că valorile se potrivesc așteptărilor.


```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Create DataFrame from previous exercises
produse = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam']
preturi = np.random.randint(50, 3000, size=5)
stocuri = np.random.randint(5, 50, size=5)

df = pd.DataFrame({
    'Produs': produse,
    'Pret': preturi,
    'Stoc': stocuri
})

print("="*60)
print("DATAFRAME INIȚIAL")
print("="*60)
print(df)

# ==================== EXERCISE 4: ADD TVA COLUMN ====================
print("\n" + "="*60)
print("EXERCIȚIUL 4: Calculare TVA (19%)")
print("="*60)

# Calculate TVA (19% of price) - VECTORIZED operation
df['TVA'] = df['Pret'] * 0.19

# Round to 2 decimals
df['TVA'] = df['TVA'].round(2)

print("\nDataFrame cu coloana TVA adăugată:")
print(df.head())

# Show the calculation for first product
print("\n📊 Exemplu calcul TVA pentru primul produs:")
print(f"   Produs: {df.iloc[0]['Produs']}")
print(f"   Preț: {df.iloc[0]['Pret']} RON")
print(f"   TVA (19%): {df.iloc[0]['TVA']} RON")
print(f"   Preț total: {df.iloc[0]['Pret'] + df.iloc[0]['TVA']:.2f} RON")

# Verify vectorization worked on all rows
print(f"\n✓ Operația vectorizată aplicată pe toate cele {len(df)} produse simultan!")
print(f"✓ Fără bucle for - Pandas face totul automat!")

# Optional: Add total price column
df['Pret_Total'] = df['Pret'] + df['TVA']
df['Pret_Total'] = df['Pret_Total'].round(2)

print("\n" + "="*60)
print("DATAFRAME FINAL CU TVA ȘI PREȚ TOTAL")
print("="*60)
print(df)
print("\n" + "="*60)
```
