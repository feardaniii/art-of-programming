### Exercițiul 1 – Set de vânzări cu index temporal
Extinde contextul comercial început anterior și construiește un `DataFrame` cu date zilnice generate prin `pd.date_range` (de exemplu, `2023-01-01` → `2023-03-31`). Include coloanele `Data`, `Oraș`, `Produs` și `Vânzări`, setează data drept index (`DatetimeIndex`), verifică tipurile pentru a confirma `datetime64[ns]` și inspectează primele și ultimele cinci rânduri.


```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

print("="*70)
print("EXERCIȚIUL 1: Set de Vânzări cu Index Temporal")
print("="*70)

# ==================== STEP 1: Generate Date Range ====================
print("\n📅 STEP 1: Generare date (2023-01-01 până la 2023-03-31)")
print("="*70)

# Create daily dates from Jan 1 to Mar 31, 2023
dates = pd.date_range(start='2023-01-01', end='2023-03-31', freq='D')

print(f"✓ Generate {len(dates)} zile")
print(f"  Prima dată: {dates[0]}")
print(f"  Ultima dată: {dates[-1]}")
print(f"  Tip: {type(dates)}")

# ==================== STEP 2: Generate Random Data ====================
print("\n📊 STEP 2: Generare date random pentru fiecare zi")
print("="*70)

# Lists of possible values
orase = ['București', 'Cluj', 'Timișoara', 'Iași', 'Brașov']
produse = ['Laptop', 'Monitor', 'Tastatura', 'Mouse', 'Webcam']

# For each date, randomly pick a city and product
n_days = len(dates)

# Random cities (one per day)
orase_random = np.random.choice(orase, size=n_days)

# Random products (one per day)
produse_random = np.random.choice(produse, size=n_days)

# Random sales numbers (between 50 and 500)
vanzari_random = np.random.randint(50, 500, size=n_days)

print(f"✓ {n_days} înregistrări generate")
print(f"  Orașe disponibile: {orase}")
print(f"  Produse disponibile: {produse}")
print(f"  Interval vânzări: 50-500 unități")

# ==================== STEP 3: Create DataFrame ====================
print("\n🏗️ STEP 3: Creare DataFrame")
print("="*70)

# Create the DataFrame
df = pd.DataFrame({
    'Data': dates,
    'Oras': orase_random,
    'Produs': produse_random,
    'Vanzari': vanzari_random
})

print("DataFrame inițial (cu Data ca și coloană normală):")
print(df.head())

print(f"\n✓ Shape: {df.shape}")
print(f"✓ Coloane: {list(df.columns)}")

# ==================== STEP 4: Set DatetimeIndex ====================
print("\n🎯 STEP 4: Setare Data ca index (DatetimeIndex)")
print("="*70)

# Set 'Data' as the index
df = df.set_index('Data')

print("DataFrame cu DatetimeIndex:")
print(df.head())

print(f"\n✓ Index type: {type(df.index)}")
print(f"✓ Index name: {df.index.name}")

# ==================== STEP 5: Verify Data Types ====================
print("\n🔍 STEP 5: Verificare tipuri de date")
print("="*70)

print("\nTipuri de date pentru fiecare coloană:")
print(df.dtypes)

print("\nDetalii despre index:")
print(f"  Tip index: {type(df.index)}")
print(f"  Primul index: {df.index[0]}")
print(f"  Ultimul index: {df.index[-1]}")
print(f"  Frecvență: {df.index.freq}")  # Should be 'D' for daily

# ==================== STEP 6: Inspect First and Last Rows ====================
print("\n📋 STEP 6: Inspecție prime și ultime 5 rânduri")
print("="*70)

print("\nPrimele 5 rânduri (.head()):")
print(df.head())

print("\nUltimele 5 rânduri (.tail()):")
print(df.tail())

# ==================== STEP 7: Additional Verification ====================
print("\n✅ STEP 7: Verificări suplimentare")
print("="*70)

print(f"\n📊 Statistici generale:")
print(f"  Total înregistrări: {len(df)}")
print(f"  Perioada: {df.index.min()} până la {df.index.max()}")
print(f"  Orașe unice: {df['Oras'].nunique()} - {df['Oras'].unique()}")
print(f"  Produse unice: {df['Produs'].nunique()} - {df['Produs'].unique()}")

print(f"\n📈 Statistici vânzări:")
print(f"  Minim: {df['Vanzari'].min()}")
print(f"  Maxim: {df['Vanzari'].max()}")
print(f"  Medie: {df['Vanzari'].mean():.2f}")
print(f"  Median: {df['Vanzari'].median():.2f}")

# Check for any missing values
print(f"\n🔎 Verificare valori lipsă:")
print(df.isnull().sum())
print("✓ Nicio valoare lipsă!" if df.isnull().sum().sum() == 0 else "⚠️ Există valori lipsă!")

# ==================== BONUS: Date-based Indexing Examples ====================
print("\n🎁 BONUS: Exemple de indexare bazată pe dată")
print("="*70)

# FIXED: Use .loc[] for all date slicing
print("\n1. Toate vânzările din Ianuarie:")
print(df.loc['2023-01'].head())

print("\n2. O săptămână specifică (Jan 1-7):")
print(df.loc['2023-01-01':'2023-01-07'])

print("\n3. Toate datele după 15 martie:")
print(df.loc['2023-03-15':].head())

# Additional useful examples
print("\n4. O singură zi (3 ianuarie):")
print(df.loc['2023-01-03'])

print("\n5. Luna februarie:")
print(df.loc['2023-02'].head())
```

### Exercițiul 4 – Strategie pentru valori lipsă
Simulează erori de raportare introducând deliberat `NaN` în coloana `Vânzări`, apoi cuantifică lipsurile cu `df.isnull().sum()` și procentul aferent pe coloană. Testează `dropna()` versus `fillna()` (de exemplu, folosind media sau mediana per pereche `Oraș`–`Produs`), notează într-un comentariu avantajele fiecărei abordări și păstrează varianta preferată pentru pașii următori.


```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

print("="*70)
print("EXERCIȚIUL 4: Strategie pentru Valori Lipsă")
print("="*70)

# ==================== STEP 0: Create the Dataset (from Ex 1) ====================
print("\n📊 STEP 0: Creare dataset inițial")
print("="*70)

# Create the time series dataset
dates = pd.date_range(start='2023-01-01', end='2023-03-31', freq='D')
n_days = len(dates)

orase = ['București', 'Cluj', 'Timișoara', 'Iași', 'Brașov']
produse = ['Laptop', 'Monitor', 'Tastatura', 'Mouse', 'Webcam']

df = pd.DataFrame({
    'Data': dates,
    'Oras': np.random.choice(orase, size=n_days),
    'Produs': np.random.choice(produse, size=n_days),
    'Vanzari': np.random.randint(50, 500, size=n_days)
})

df = df.set_index('Data')

print(f"✓ Dataset creat: {df.shape[0]} rânduri, {df.shape[1]} coloane")
print(f"✓ Nicio valoare lipsă inițial")
print("\nPrimele 5 rânduri:")
print(df.head())

# ==================== STEP 1: Simulate Missing Values ====================
print("\n" + "="*70)
print("🔧 STEP 1: Simulare valori lipsă (NaN)")
print("="*70)

# Create a copy to preserve original
df_messy = df.copy()

# Randomly introduce NaN in ~10% of Vanzari values
n_missing = int(len(df_messy) * 0.10)  # 10% missing
missing_indices = np.random.choice(df_messy.index, size=n_missing, replace=False)
df_messy.loc[missing_indices, 'Vanzari'] = np.nan

print(f"✓ Am introdus {n_missing} valori lipsă (~10% din date)")
print("\nPrimele rânduri cu date lipsă:")
print(df_messy[df_messy['Vanzari'].isnull()].head())

# ==================== STEP 2: Quantify Missing Values ====================
print("\n" + "="*70)
print("📊 STEP 2: Cuantificare valori lipsă")
print("="*70)

# Count missing values per column
missing_counts = df_messy.isnull().sum()
print("\nValori lipsă per coloană:")
print(missing_counts)

# Percentage of missing values
missing_percentage = (df_messy['Vanzari'].isnull().sum() / len(df_messy)) * 100
print(f"\nProcent valori lipsă în Vanzari: {missing_percentage:.2f}%")

# Total missing
total_missing = df_messy['Vanzari'].isnull().sum()
print(f"Total valori lipsă: {total_missing} din {len(df_messy)}")

# ==================== STEP 3: Strategy 1 - dropna() ====================
print("\n" + "="*70)
print("🗑️ STEP 3: STRATEGIE 1 - dropna() (Elimină rândurile)")
print("="*70)

df_dropped = df_messy.dropna()

print(f"\nÎnainte: {len(df_messy)} rânduri")
print(f"După dropna(): {len(df_dropped)} rânduri")
print(f"Rânduri eliminate: {len(df_messy) - len(df_dropped)}")

print(f"\n📊 Statistici după dropna():")
print(f"  Minim: {df_dropped['Vanzari'].min():.0f}")
print(f"  Maxim: {df_dropped['Vanzari'].max():.0f}")
print(f"  Medie: {df_dropped['Vanzari'].mean():.2f}")
print(f"  Median: {df_dropped['Vanzari'].median():.2f}")

print("\n✅ PRO-uri dropna():")
print("  + Date 100% reale (nu facem presupuneri)")
print("  + Simplu și rapid")
print("  + Nu introduce bias")

print("\n❌ CONTRA dropna():")
print("  - Pierdem informație (orașe, produse)")
print("  - Dacă avem multe NaN, pierdem mult din date")
print("  - Poate crea găuri în seria temporală")

# ==================== STEP 4: Strategy 2 - fillna() with Mean ====================
print("\n" + "="*70)
print("🔄 STEP 4: STRATEGIE 2 - fillna() cu Media")
print("="*70)

# Fill with overall mean
df_filled_mean = df_messy.copy()
mean_vanzari = df_messy['Vanzari'].mean()
df_filled_mean['Vanzari'] = df_filled_mean['Vanzari'].fillna(mean_vanzari)

print(f"\nMedia vânzărilor (pentru fill): {mean_vanzari:.2f}")
print(f"\nÎnainte: {df_messy['Vanzari'].isnull().sum()} NaN")
print(f"După fillna(mean): {df_filled_mean['Vanzari'].isnull().sum()} NaN")

print(f"\n📊 Statistici după fillna(mean):")
print(f"  Minim: {df_filled_mean['Vanzari'].min():.0f}")
print(f"  Maxim: {df_filled_mean['Vanzari'].max():.0f}")
print(f"  Medie: {df_filled_mean['Vanzari'].mean():.2f}")
print(f"  Median: {df_filled_mean['Vanzari'].median():.2f}")

# ==================== STEP 5: Strategy 3 - fillna() with Median ====================
print("\n" + "="*70)
print("🔄 STEP 5: STRATEGIE 3 - fillna() cu Mediana")
print("="*70)

# Fill with median
df_filled_median = df_messy.copy()
median_vanzari = df_messy['Vanzari'].median()
df_filled_median['Vanzari'] = df_filled_median['Vanzari'].fillna(median_vanzari)

print(f"\nMediana vânzărilor (pentru fill): {median_vanzari:.2f}")
print(f"\nÎnainte: {df_messy['Vanzari'].isnull().sum()} NaN")
print(f"După fillna(median): {df_filled_median['Vanzari'].isnull().sum()} NaN")

print(f"\n📊 Statistici după fillna(median):")
print(f"  Minim: {df_filled_median['Vanzari'].min():.0f}")
print(f"  Maxim: {df_filled_median['Vanzari'].max():.0f}")
print(f"  Medie: {df_filled_median['Vanzari'].mean():.2f}")
print(f"  Median: {df_filled_median['Vanzari'].median():.2f}")

print("\n✅ PRO-uri fillna():")
print("  + Păstrăm toate rândurile (nu pierdem date)")
print("  + Menține dimensiunea dataset-ului")
print("  + Păstrăm informația despre orașe/produse")

print("\n❌ CONTRA fillna():")
print("  - Introducem presupuneri (date artificiale)")
print("  - Poate distorsiona distribuția")
print("  - Mean vs Median - care alegem?")

# ==================== STEP 6: Advanced - fillna() per Group ====================
print("\n" + "="*70)
print("🎯 STEP 6: STRATEGIE AVANSATĂ - fillna() per Oraș~Produs")
print("="*70)

# Fill with mean per Oras-Produs combination
df_filled_group = df_messy.copy()

# Calculate mean per group
group_means = df_messy.groupby(['Oras', 'Produs'])['Vanzari'].transform('mean')

# Fill NaN with group mean
df_filled_group['Vanzari'] = df_filled_group['Vanzari'].fillna(group_means)

# If still NaN (group had no valid values), fill with overall mean
df_filled_group['Vanzari'] = df_filled_group['Vanzari'].fillna(mean_vanzari)

print(f"\n✓ Am umplut NaN cu media per combinație Oraș-Produs")
print(f"✓ Dacă grupul nu are date, folosim media globală")
print(f"\nNaN rămase: {df_filled_group['Vanzari'].isnull().sum()}")

print(f"\n📊 Statistici după fillna(group mean):")
print(f"  Minim: {df_filled_group['Vanzari'].min():.0f}")
print(f"  Maxim: {df_filled_group['Vanzari'].max():.0f}")
print(f"  Medie: {df_filled_group['Vanzari'].mean():.2f}")
print(f"  Median: {df_filled_group['Vanzari'].median():.2f}")

print("\n🌟 De ce e mai bun group mean?")
print("  + Vânzările Laptop în București ≠ Mouse în Iași")
print("  + Media per grup e mai relevantă decât media globală")
print("  + Mai aproape de realitate")

# ==================== STEP 7: Comparison Table ====================
print("\n" + "="*70)
print("📊 STEP 7: COMPARAȚIE STRATEGII")
print("="*70)

comparison = pd.DataFrame({
    'Strategie': ['Original (cu NaN)', 'dropna()', 'fillna(mean)', 'fillna(median)', 'fillna(group)'],
    'Rânduri': [
        len(df_messy),
        len(df_dropped),
        len(df_filled_mean),
        len(df_filled_median),
        len(df_filled_group)
    ],
    'NaN Count': [
        df_messy['Vanzari'].isnull().sum(),
        df_dropped['Vanzari'].isnull().sum(),
        df_filled_mean['Vanzari'].isnull().sum(),
        df_filled_median['Vanzari'].isnull().sum(),
        df_filled_group['Vanzari'].isnull().sum()
    ],
    'Mean': [
        df_messy['Vanzari'].mean(),
        df_dropped['Vanzari'].mean(),
        df_filled_mean['Vanzari'].mean(),
        df_filled_median['Vanzari'].mean(),
        df_filled_group['Vanzari'].mean()
    ]
})

print("\n" + comparison.to_string(index=False))

# ==================== STEP 8: Recommendation ====================
print("\n" + "="*70)
print("💡 STEP 8: RECOMANDARE FINALĂ")
print("="*70)

print("\n🎯 CÂND SĂ FOLOSEȘTI FIECARE STRATEGIE:")

print("\n1️⃣ dropna() - Folosește când:")
print("   • Ai < 5% valori lipsă")
print("   • Datele lipsă sunt random (nu sistematice)")
print("   • Ai suficiente date și poți pierde câteva rânduri")

print("\n2️⃣ fillna(mean/median) - Folosește când:")
print("   • Ai 5-20% valori lipsă")
print("   • Vrei să păstrezi toate rândurile")
print("   • Datele sunt relativ uniforme")

print("\n3️⃣ fillna(group mean) - Folosește când:")
print("   • Ai 5-30% valori lipsă")
print("   • Există grupuri clare în date (oraș, produs)")
print("   • Vrei cea mai bună aproximare")
print("   ⭐ RECOMANDAT pentru acest dataset!")

print("\n❌ NU folosi fillna() când:")
print("   • Ai > 30% valori lipsă (date prea corupte)")
print("   • Valorile lipsă nu sunt random (bias sistematic)")

print("\n🏆 PENTRU ACEST EXERCIȚIU:")
print(f"   Cu {missing_percentage:.1f}% valori lipsă, recomand: fillna(group mean)")
print("   Motivație: Păstrăm toate datele + respectăm contextul Oraș-Produs")

print("\n" + "="*70)
print("✅ EXERCIȚIUL 4 COMPLETAT!")
print("="*70)

# Save the cleaned dataset for next exercises
df_clean = df_filled_group.copy()
print(f"\n💾 Dataset curățat salvat în variabila 'df_clean'")
print(f"   {len(df_clean)} rânduri, 0 NaN-uri, gata pentru vizualizare!")
```
