## Ex. 1 — Test **Singleton**
- Creează o clasă `Logger` implementată ca **Singleton** (o singură instanță în tot programul).
- Metodă minimă: `log(mesaj: str)` care afișează sau reține mesajul.
- Creează două variabile `log1` și `log2` și demonstrează că referă **aceeași** instanță (ex.: `id(log1) == id(log2)`).
- Nu folosi soluții externe; implementează tu mecanismul Singleton.


```python
class Logger:
    # Această variabilă de clasă va păstra instanța unică
    _instance = None

    def __new__(cls):
        # Dacă _instance este None, înseamnă că nu am creat încă obiectul
        if cls._instance is None:
            # Creăm instanța folosind metoda clasei părinte (object)
            cls._instance = super(Logger, cls).__new__(cls)
            # Putem inițializa aici o listă de mesaje dacă vrem să le reținem
            cls._instance.istoric_mesaje = []
        
        # Returnăm instanța existentă (sau cea tocmai creată)
        return cls._instance

    def log(self, mesaj: str):
        print(f"[LOG]: {mesaj}")
        # Opțional: salvăm mesajul în instanță
        self.istoric_mesaje.append(mesaj)

# --- Demonstrația cerută ---

# Creăm două variabile care încearcă să instanțieze clasa
log1 = Logger()
log2 = Logger()

# 1. Verificăm dacă referă aceeași instanță folosind id()
print(f"ID log1: {id(log1)}")
print(f"ID log2: {id(log2)}")

if id(log1) == id(log2):
    print("Succes: log1 și log2 sunt aceeași instanță (Singleton).")
else:
    print("Eroare: Instanțele sunt diferite.")

# 2. Testăm funcționalitatea metodei log
log1.log("Prima acțiune a utilizatorului.")
log2.log("A doua acțiune a utilizatorului.")

# Demonstram că datele sunt comune (dacă am salvat în listă)
print(f"Istoric mesaje accesat prin log1: {log1.istoric_mesaje}")
```

## Ex. 2 — Test **Factory**
- Implementează o **Factory** (de ex. `ProdusFactory`) care creează obiecte diferite în funcție de un parametru `tip` (ex.: `"telefon"`, `"laptop"`).
- Fiecare tip are o clasă cu metoda `descriere()`.
- Apelează `ProdusFactory.creeaza_produs("telefon")` și afișează descrierea.


```python
# 1. Definim clasele pentru produsele specifice
class Telefon:
    def descriere(self):
        return "Acesta este un smartphone cu ecran tactil."

class Laptop:
    def descriere(self):
        return "Acesta este un laptop performant pentru lucru și gaming."

# 2. Implementăm clasa Factory
class ProdusFactory:
    @staticmethod
    def creeaza_produs(tip_produs: str):
        """
        Această metodă decide ce obiect să creeze în funcție de parametrul primit.
        """
        if tip_produs.lower() == "telefon":
            return Telefon()
        elif tip_produs.lower() == "laptop":
            return Laptop()
        else:
            # Opțional: putem returna None sau ridica o eroare dacă tipul nu există
            raise ValueError(f"Tipul de produs '{tip_produs}' nu este recunoscut.")

# --- Demonstrația cerută ---

# Folosim fabrica pentru a crea obiecte fără a apela direct Telefon() sau Laptop()
produs1 = ProdusFactory.creeaza_produs("telefon")
produs2 = ProdusFactory.creeaza_produs("laptop")

# Afișăm descrierile
print(f"Produs 1: {produs1.descriere()}")
print(f"Produs 2: {produs2.descriere()}")

# Exemplu de gestionare a unui tip necunoscut (opțional)
try:
    produs3 = ProdusFactory.creeaza_produs("tableta")
except ValueError as e:
    print(f"\nEroare așteptată: {e}")
```

## Ex. 3 — Test **Builder**
- Creează un `LaptopBuilder` cu metode de configurare (ex.: `.cu_ram(int)`, `.cu_ssd(int)`, `.cu_gpu(str)`).
- Metoda finală `.construieste()` întoarce un obiect `Laptop` cu o metodă `descriere()`.
- Construiește un laptop personalizat și afișează descrierea.


```python
# 1. Clasa Produs - obiectul final pe care dorim să îl obținem
class Laptop:
    def __init__(self, ram, ssd, gpu):
        self.ram = ram
        self.ssd = ssd
        self.gpu = gpu

    def descriere(self):
        return f"Laptop configurat cu: {self.ram}GB RAM, {self.ssd}GB SSD și placă video {self.gpu}."

# 2. Clasa Builder
class LaptopBuilder:
    def __init__(self):
        # Setăm niște valori implicite (default)
        self.ram = 8
        self.ssd = 256
        self.gpu = "Integrată"

    def cu_ram(self, ram: int):
        self.ram = ram
        return self  # Returnăm self pentru a permite "method chaining"

    def cu_ssd(self, ssd: int):
        self.ssd = ssd
        return self

    def cu_gpu(self, gpu: str):
        self.gpu = gpu
        return self

    def construieste(self):
        # Metoda finală care instanțiază obiectul Laptop cu datele colectate
        return Laptop(self.ram, self.ssd, self.gpu)

# --- Demonstrația cerută ---

# Putem folosi "Method Chaining" (apelarea metodelor una după alta pe aceeași linie)
builder = LaptopBuilder()
laptop_personalizat = (builder
                       .cu_ram(32)
                       .cu_ssd(1024)
                       .cu_gpu("NVIDIA RTX 4070")
                       .construieste())

# Afișăm descrierea
print(laptop_personalizat.descriere())

# Putem crea și un laptop cu setări minime (folosind valorile default pentru restul)
laptop_office = LaptopBuilder().cu_ram(16).construieste()
print(laptop_office.descriere())
```
